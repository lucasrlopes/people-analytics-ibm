```python
import pandas as pd

pd.set_option("display.max_columns", None)

df = pd.read_csv("data/WA_Fn-UseC_-HR-Employee-Attrition.csv")
df.shape
```

```python
df.info()
```

```python
print("Valores vazios:", df.isna().sum().sum())
print("Linhas duplicadas:", df.duplicated().sum())
print("ID de funcionário único?", df["EmployeeNumber"].is_unique)
```

```python
constantes = df.columns[df.nunique() == 1].tolist()
print("Colunas com um valor só:", constantes)

df = df.drop(columns=constantes)
df.shape
```

```python
pd.DataFrame({
    "funcionarios": df["Attrition"].value_counts(),
    "percentual": (df["Attrition"].value_counts(normalize=True) * 100).round(1)
})
```
